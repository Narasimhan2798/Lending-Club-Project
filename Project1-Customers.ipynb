```python
customer_schema = '''member_id string, emp_title string, emp_length string, home_ownership string, annual_inc float, addr_state string, zip_code string, country string,
grade string, sub_grade string, verification_status string, tot_hi_cred_lim float, application_type string, annual_inc_joint float, verification_status_joint string'''
```

```python
customers_raw_df = spark.read \
.format("csv") \
.option("header","true") \
.schema(customer_schema) \
.load("/user/itv026931/lendingclubproject/raw/customer_data_csv")
```

```python
customers_raw_df.printSchema()
```

RENAMING THE COLUMNS

```python
customers_df_renamed = customers_raw_df.withColumnRenamed("annual_inc","annual_income").withColumnRenamed("addr_state","address_state").withColumnRenamed("zip_code","address_zipcode") \
.withColumnRenamed("country","address_country").withColumnRenamed("tot_hi_cred_lim","total_high_credit_limit").withColumnRenamed("annual_inc_joint","join_annual_income")
```

```python
customers_df_renamed
```

INSERTING A COLUMN WITH CURRENT TIMESTAMP

```python
customers_df_ingestd = customers_df_renamed.withColumn("ingest_date",current_timestamp())
```

REMOVE COMPLETE DUPLICATE ROWS

```python
customers_df_ingestd.count()
```

```python
customers_distinct = customers_df_ingestd.distinct()
```

```python
customers_distinct.count()
```

```python
customers_distinct.createOrReplaceTempView("customers")
```

```python
spark.sql("select * from customers")
```

REMOVING THE ROWS WHERE 'ANNUAL INCOME' is NULL

```python
customers_income_filtered = customers_distinct.filter("annual_income is not null")
```

```python
customers_income_filtered.createOrReplaceTempView("customers")
```

```python
spark.sql("Select count(*) from customers where annual_income is null")
```

REMOVING THE STRING CHARACTERS FROM "EMP_LENGTH" and MAKING IT AS A NUMERICAL VALUE

```python
customers_emplength_cleaned = customers_income_filtered.withColumn("emp_length",regexp_replace(col("emp_length"),"\D",""))
```

```python
customers_emplength_cleaned.createOrReplaceTempView("customers")
```

CONVERTING THE COLUMN FROM 1 DATATYPE TO ANOTHER

```python
customers_emplength_casted = customers_emplength_cleaned.withColumn("emp_length",customers_emplength_cleaned.emp_length.cast("int"))
```

```python
customers_emplength_casted.printSchema()
```

```python
customers_emplength_casted.createOrReplaceTempView("customers")
```

MAKING THE "EMP_LENGTH" TO CLEANED WAY BY REMOVING NULL VALUES

```python
avg_length = customers_emplength_casted.agg(avg(col("emp_length")).alias("avg_emp_length")).collect()
```

```python
avg_emp_length = spark.sql("select floor(avg(emp_length)) as avg_emp_length from customers").collect()
```

```python
print(avg_emp_length)
```

```python
avg_emp_duration = avg_emp_length[0][0]
```

```python
customers_emplength_replaced = customers_emplength_casted.na.fill(avg_emp_duration,subset=['emp_length'])
```

```python
customers_emplength_replaced.createOrReplaceTempView("customers")
```

```python
spark.sql("select distinct(emp_length) as emp_length from customers")
```

Clean the address_state [it should be 2 characters only], replace all others with NA

```python
spark.sql("select distinct address_state,length(address_state) as length from customers")
```

```python
customers_state_cleaned = customers_emplength_replaced.withColumn("address_state",when(length(col("address_state")) > 2, "NA").otherwise(col("address_state")))
```

```python
customers_state_cleaned.createOrReplaceTempView("Customers")
```

```python
spark.sql("select distinct address_state from customers")
```

```python
customers_state_cleaned.write \
.format("parquet") \
.mode("overwrite") \
.option("path","/user/itv026931/lendingclubproject/cleaned/customers_parquet") \
.save()
```
