```python
spark.sql("create database itv026931_lending_club")
```

```python
spark.sql("""
create EXTERNAL table itv026931_lending_club.customers(member_id string,emp_title
string,emp_length int, home_ownership string,annual_income
float,address_state string,address_zipcode string,address_country
string,grade string,sub_grade string,verification_status
string,total_high_credit_limit float,application_type
string,join_annual_income float,verification_status_joint string,ingest_date
timestamp) stored as parquet
LOCATION '/user/itv026931/lendingclubproject/cleaned/customers_parquet'
""")
```

```python
spark.sql("""
create EXTERNAL table itv026931_lending_club.loans(loan_id
string,member_id string,loan_amount float,funded_amount
float,loan_term_years integer,interest_rate float,monthly_installment
float,issue_date string,loan_status string,loan_purpose string,loan_title
string,ingest_date timestamp) stored as parquet
LOCATION '/user/itv026931/lendingclubproject/cleaned/loans_parquet'
""")
```

```python
spark.sql("""
create EXTERNAL table itv026931_lending_club.loans_repayments(loan_id
string,total_principal_received float,total_interest_received
float,total_late_fee_received float,total_payment_received
float,last_payment_amount float,last_payment_date string,next_payment_date
string,ingest_date timestamp) stored as parquet
LOCATION '/user/itv026931/lendingclubproject/cleaned/loan_repayment_parquet'
""")
```

```python
spark.sql("""
create EXTERNAL table itv026931_lending_club.loans_defaulters_delinq(member_id
string,delinq_2yrs integer,delinq_amnt float,mths_since_last_delinq integer) stored as parquet
LOCATION '/user/itv026931/lendingclubproject/cleaned/loans_defaulters_delinq_parquet'
""")
```

```python
spark.sql("""
create EXTERNAL table itv026931_lending_club.loans_defaulters_detail_rec_enq_nar(member_id string,
pub_rec integer, pub_rec_bankruptcies integer, inq_last_6mths integer) stored as parquet
LOCATION '/user/itv026931/lendingclubproject/cleaned/loans_def_detail_records_enq_df_parquet'
""")
```

```python
spark.sql("""
CREATE or REPLACE view itv026931_lending_club.customers_loan_v as SELECT
l.loan_id,
c.member_id,
c.emp_title,
c.emp_length,
c.home_ownership,
c.annual_income,
c.address_state,
c.address_zipcode,
c.address_country,
c.grade,
c.sub_grade,
c.verification_status,
c.total_high_credit_limit,
c.application_type,
c.join_annual_income,
c.verification_status_joint,
l.loan_amount,
l.funded_amount,
l.loan_term_years,
l.interest_rate,
l.monthly_installment,
l.issue_date,
l.loan_status,
l.loan_purpose,
r.total_principal_received,
r.total_interest_received,
r.total_late_fee_received,
r.last_payment_date,
r.next_payment_date,
d.delinq_2yrs,
d.delinq_amnt,
d.mths_since_last_delinq,
e.pub_rec,
e.pub_rec_bankruptcies,
e.inq_last_6mths
from itv026931_lending_club.customers as c
left join itv026931_lending_club.loans as l on c.member_id = l.member_id
left join itv026931_lending_club.loans_repayments as r on l.loan_id = r.loan_id
left join itv026931_lending_club.loans_defaulters_deling as d on c.member_id = d.member_id
left join itv026931_lending_club.loans_defaulters_detail_rec_enq_nar as e on c.member_id = e.member_id
""")
```

```python
spark.sql("select * from itv026931_lending_club.customers_loan_v")
```

```python
spark.sql("""
create table itv026931_lending_club.customers_loan_t as SELECT
l.loan_id,
c.member_id,
c.emp_title,
c.emp_length,
c.home_ownership,
c.annual_income,
c.address_state,
c.address_zipcode,
c.address_country,
c.grade,
c.sub_grade,
c.verification_status,
c.total_high_credit_limit,
c.application_type,
c.join_annual_income,
c.verification_status_joint,
l.loan_amount,
l.funded_amount,
l.loan_term_years,
l.interest_rate,
l.monthly_installment,
l.issue_date,
l.loan_status,
l.loan_purpose,
r.total_principal_received,
r.total_interest_received,
r.total_late_fee_received,
r.last_payment_date,
r.next_payment_date,
d.delinq_2yrs,
d.delinq_amnt,
d.mths_since_last_delinq,
e.pub_rec,
e.pub_rec_bankruptcies,
e.inq_last_6mths
from itv026931_lending_club.customers as c
left join itv026931_lending_club.loans as l on c.member_id = l.member_id
left join itv026931_lending_club.loans_repayments as r on l.loan_id = r.loan_id
left join itv026931_lending_club.loans_defaulters_deling as d on c.member_id = d.member_id
left join itv026931_lending_club.loans_defaulters_detail_rec_enq_nar as e on c.member_id = e.member_id
""")
```
